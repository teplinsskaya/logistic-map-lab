## Easy level, задание 1

**Утверждение:**  
Для логистического отображения $x_{n+1} = r x_n (1 - x_n)$ при $r \in (0; 1]$ и $0 < x_0 < 1$ выполняется:  
$\forall n \in \mathbb{N}: \quad 0 < x_n < 1$

**Доказательство** (метод математической индукции):

**1. База индукции**  
При $n = 0$ условие $0 < x_0 < 1$ дано. База выполнена.

**2. Предположение индукции**  
Предположим, что для некоторого $n$ выполняется:
$0 < x_n < 1$

**3. Шаг индукции**  
Требуется доказать, что:
$0 < x_{n+1} < 1$
где $x_{n+1} = r \, x_n \, (1 - x_n)$

**а) Докажем, что $x_{n+1} > 0$:**  
Из предположения: $x_n > 0$ и $x_n < 1$ ⇒ $1 - x_n > 0$  
Так как $r > 0$ (по уточнённому условию), все три множителя **строго положительны**. Следовательно,
$x_{n+1} = r \, x_n \, (1 - x_n) > 0$

**б) Докажем, что $x_{n+1} < 1$:**  
Рассмотрим функцию $f(x) = r x (1 - x)$ на интервале $x \in (0, 1)$  
Это квадратичная функция:
$f(x) = -r x^2 + r x$

Её график — парабола с ветвями вниз (коэффициент при $x^2$ равен $-r < 0$)

Вершина параболы:
$x_{\text{в}} = \frac{-r}{2 \cdot (-r)} = \frac{1}{2}$

$f\!\left(\frac{1}{2}\right) = r \cdot \frac{1}{2} \cdot \left(1 - \frac{1}{2}\right) = \frac{r}{4}$

Поскольку $r \le 1$, имеем $f(1/2) \le 1/4 < 1$

Значит, максимальное значение $f(x)$ на $(0, 1)$ меньше 1, и для любого $x_n \in (0, 1)$:
$x_{n+1} = f(x_n) < 1$

**4. Заключение**  
Из $0 < x_n < 1$ следует $0 < x_{n+1} < 1$ при $r \in (0; 1]$  
По индукции это верно для всех $n \in \mathbb{N}$


**Доказательство завершено для $r \in (0; 1]$**  
При $r = 0$ последовательность вырождается в нулевую, что является особым случаем

```python
# графики для логистического отображения

import numpy as np
import matplotlib.pyplot as plt

# логистическое отображение
def logistic(x, r):
    return r * x * (1 - x)

# создаём значения x от 0 до 1
x = np.linspace(0, 1, 500)

# значения r для построения (разные диапазоны)
r_values = [0.5, 1.0, 2.0, 3.0, 3.5]

# настраиваем график
plt.figure(figsize=(10, 6))
plt.title('Задание 2. Логистическое отображение: $x_n = r x_{n-1} (1 - x_{n-1})$', fontsize=14)
plt.xlabel('$x_{n-1}$', fontsize=12)
plt.ylabel('$x_n$', fontsize=12)
plt.grid(True, alpha=0.3)

# линия y = x для сравнения
plt.plot(x, x, 'k--', alpha=0.5, label='$y = x$')

# строим графики для разных r
for r in r_values:
    y = logistic(x, r)
    plt.plot(x, y, label=f'$r = {r}$')

plt.legend()
plt.xlim(0, 1)
plt.ylim(0, 1)
plt.show()
```

## Задание 1: Неподвижные точки логистического отображения

### 1.1 Нахождение неподвижных точек
Неподвижная точка $x^*$ удовлетворяет уравнению:
$x^* = f(x^*) = r x^* (1 - x^*)$

Решаем:
$x^* = r x^* (1 - x^*)$
$x^* - r x^* + r (x^*)^2 = 0$
$x^*(1 - r + r x^*) = 0$

Получаем два решения:
1. $x_1^* = 0$
2. $1 - r + r x^* = 0$ ⇒ $x_2^* = \frac{r-1}{r}$ (при $r \neq 0$)

**Ответ:** Неподвижные точки логистического отображения:
- $x_1^* = 0$ (существует при всех $r$)
- $x_2^* = 1 - \frac{1}{r}$ (существует при $r \neq 0$)

### 1.2 Количество неподвижных точек
- При $r = 0$: только $x^* = 0$ (одна точка)
- При $0 < r < 1$: две точки, но $x_2^* < 0$ (не входит в [0,1])
- При $r = 1$: $x_1^* = 0$, $x_2^* = 0$ (точки совпадают)
- При $r > 1$: две различные точки в [0,1]

### 1.3 Максимальное количество неподвижных точек
Логистическое отображение — квадратичная функция ($f(x) = -r x^2 + r x$).  
Уравнение $x = f(x)$ — квадратное уравнение, которое имеет не более 2 корней.  
**Ответ:** Максимально 2 неподвижные точки.

## Задание 2: Монотонность и предел при r ∈ (0; 1]

### 2.1 Доказательство монотонного убывания

**Дано:** Логистическое отображение $x_{n+1} = r x_n (1 - x_n)$, где $r \in (0; 1]$, $x_0 \in (0; 1)$.

**Доказательство:**

1. Рассмотрим разность двух последовательных членов:
   $x_{n+1} - x_n = r x_n (1 - x_n) - x_n = x_n [r(1 - x_n) - 1]$

2. Так как $x_n > 0$ (доказано в Easy level), знак разности определяется выражением в скобках:
   $r(1 - x_n) - 1$

3. Поскольку $r \leq 1$ и $0 < 1 - x_n < 1$, имеем:
   $r(1 - x_n) \leq 1 \cdot (1 - x_n) < 1$
   
   Следовательно:
   $r(1 - x_n) - 1 < 0$

4. Таким образом:
   $x_{n+1} - x_n = x_n [r(1 - x_n) - 1] < 0$
   $x_{n+1} < x_n$ для всех $n$

**Вывод:** Последовательность $\{x_n\}$ строго монотонно убывает.

### 2.2 Существование предела

**Доказательство:**

1. Последовательность $\{x_n\}$ монотонно убывает (доказано выше).
2. Последовательность ограничена снизу нулём: $x_n > 0$ (доказано в Easy level).
3. По теореме Вейерштрасса о монотонной ограниченной последовательности, существует конечный предел:
   $\lim_{n \to \infty} x_n = L$

4. Найдём $L$, переходя к пределу в уравнении отображения:
   $L = r L (1 - L)$
   $L[1 - r(1 - L)] = 0$

5. Решения:
   - $L = 0$
   - $1 - r(1 - L) = 0 \Rightarrow L = 1 - \frac{1}{r}$

6. При $r \in (0; 1]$:
   - Второе решение $L = 1 - 1/r$ либо отрицательно ($r < 1$), либо равно 0 ($r = 1$)
   - Единственный допустимый предел в $(0, 1)$ — $L = 0$

**Вывод:** При $r \in (0; 1]$ предел существует и равен 0: $\lim_{n \to \infty} x_n = 0$

### 2.3 Графическая иллюстрация

```python
import numpy as np
import matplotlib.pyplot as plt

r_values = [0.3, 0.7, 1.0]
x0 = 0.8
iterations = 30

plt.figure(figsize=(12, 4))

for i, r in enumerate(r_values, 1):
    x_vals = [x0]
    for n in range(iterations):
        x_next = r * x_vals[-1] * (1 - x_vals[-1])
        x_vals.append(x_next)

    plt.subplot(1, 3, i)
    plt.plot(range(iterations+1), x_vals, 'bo-', markersize=3, linewidth=1)
    plt.title(f'$r = {r}$, $x_0 = {x0}$', fontsize=12)
    plt.xlabel('$n$', fontsize=11)
    plt.ylabel('$x_n$', fontsize=11)
    plt.grid(True, alpha=0.3)
    plt.ylim(0, 1)

plt.tight_layout()
plt.show()
```

## Задание 3: Подпоследовательности при r ∈ (2; 3)

### 3.1 Доказательство монотонности подпоследовательностей

**Дано:** $r \in (2; 3)$, $x_0 \in (0; 1)$, $x_{2n} > x^*$, $x_{2n+1} < x^*$, где $x^* = 1 - 1/r$ — положительная неподвижная точка.

**Доказательство для чётной подпоследовательности $\{x_{2n}\}$:**

1. Рассмотрим функцию $f^{(2)}(x) = f(f(x)) = r[r x (1 - x)][1 - r x (1 - x)]$
   Это второе итерирование отображения.

2. Производная в неподвижной точке:
   $[f^{(2)}]'(x^*) = [f'(x^*)]^2 = (2 - r)^2$
   
   При $r \in (2; 3)$: $0 < (2 - r)^2 < 1$

3. Так как $x_{2n} > x^*$ и производная $[f^{(2)}]'(x) < 1$ в окрестности $x^*$, то:
   $x_{2(n+1)} = f^{(2)}(x_{2n}) < x_{2n}$

**Вывод:** Чётная подпоследовательность $\{x_{2n}\}$ монотонно убывает.

**Доказательство для нечётной подпоследовательности $\{x_{2n+1}\}$:**

1. Аналогично, для $x_{2n+1} < x^*$:
   $x_{2(n+1)+1} = f^{(2)}(x_{2n+1}) > x_{2n+1}$
   (так как $f^{(2)}$ возрастает в окрестности $x^*$ при $f'(x^*) < 0$)

2. При $r \in (2; 3)$: $f'(x^*) = 2 - r < 0$, поэтому $f^{(2)}$ имеет положительную производную.

**Вывод:** Нечётная подпоследовательность $\{x_{2n+1}\}$ монотонно возрастает.

### 3.2 Графическая иллюстрация

```python
r = 2.7  # Выбираем r ∈ (2; 3)
x0 = 0.8  # Начальное значение > x*
iterations = 40

x_vals = [x0]
for n in range(iterations):
    x_next = r * x_vals[-1] * (1 - x_vals[-1])
    x_vals.append(x_next)

x_star = 1 - 1/r

# разделение на подпоследовательности
even_seq = x_vals[0::2]
odd_seq = x_vals[1::2]

# графики
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# 1. Полная последовательность
axes[0].plot(range(len(x_vals)), x_vals, 'bo-', markersize=3, linewidth=1)
axes[0].axhline(y=x_star, color='r', linestyle='--', label=f'$x^*$ = {x_star:.3f}')
axes[0].set_title(f'Полная последовательность, $r = {r}$')
axes[0].set_xlabel('$n$')
axes[0].set_ylabel('$x_n$')
axes[0].grid(True, alpha=0.3)
axes[0].legend()

# 2. Чётные члены
axes[1].plot(range(len(even_seq)), even_seq, 'go-', markersize=3, linewidth=1.5)
axes[1].axhline(y=x_star, color='r', linestyle='--', alpha=0.7)
axes[1].set_title('Чётные члены $x_{2k}$ (убывают)')
axes[1].set_xlabel('$k$')
axes[1].set_ylabel('$x_{2k}$')
axes[1].grid(True, alpha=0.3)

# 3. Нечётные члены
axes[2].plot(range(len(odd_seq)), odd_seq, 'mo-', markersize=3, linewidth=1.5)
axes[2].axhline(y=x_star, color='r', linestyle='--', alpha=0.7)
axes[2].set_title('Нечётные члены $x_{2k+1}$ (возрастают)')
axes[2].set_xlabel('$k$')
axes[2].set_ylabel('$x_{2k+1}$')
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Задание 4: Для отображения $g(x) = rx(1-x)(2-x)$ (N=4)

### 4.1 Неподвижные точки

Решаем $x = g(x) = rx(1-x)(2-x)$:

$x = rx(1-x)(2-x)$

1. Первое решение: $x_1^* = 0$

2. Для $x \neq 0$ делим на $x$:
   $1 = r(1-x)(2-x)$
   $1 = r(2 - 3x + x^2)$
   $rx^2 - 3rx + (2r - 1) = 0$

   Дискриминант:
   $D = 9r^2 - 4r(2r-1) = 9r^2 - 8r^2 + 4r = r^2 + 4r$

   Решения:
   $x_{2,3}^* = \frac{3r \pm \sqrt{r^2 + 4r}}{2r}$

**Ответ:** Неподвижные точки:
- $x_1^* = 0$ (всегда)
- $x_2^* = \frac{3r - \sqrt{r^2 + 4r}}{2r}$
- $x_3^* = \frac{3r + \sqrt{r^2 + 4r}}{2r}$ (при $r > 0$)

### 4.2 Диапазон $r$ для монотонной сходимости к 0

**Условия монотонной сходимости к 0:**
1. $|g'(0)| < 1$ (устойчивость нуля)
2. $g(x) < x$ для $x \in (0,1)$ (монотонность)

**Вычисляем:**
$g'(x) = r[(1-x)(2-x) - x(2-x) - x(1-x)] = r(2 - 6x + 3x^2)$
$g'(0) = 2r$

**Условие устойчивости:** $|g'(0)| < 1$ ⇒ $2r < 1$ ⇒ $r < 0.5$

**Проверка $g(x) < x$:**
При малых $x$ и $r < 0.5$:
$g(x) \approx 2rx < x$ (т.к. $2r < 1$)

**Ответ:** Последовательность монотонно сходится к 0 при $r \in [0, 0.5)$.

### 4.3 Графики зависимости $x_n$ от $n$

```python
def g(x, r):
    return r * x * (1 - x) * (2 - x)

# разные значения r
r_values = [0.3, 0.8, 1.5, 2.0, 2.5]
x0 = 0.3
iterations = 50

plt.figure(figsize=(15, 8))

for i, r in enumerate(r_values, 1):
    # генерация последовательности
    x_vals = [x0]
    for n in range(iterations):
        x_next = g(x_vals[-1], r)
        x_vals.append(x_next)

    plt.subplot(2, 3, i)
    plt.plot(range(iterations+1), x_vals, 'o-', markersize=2, linewidth=1)
    plt.title(f'$r = {r}$', fontsize=12)
    plt.xlabel('$n$', fontsize=11)
    plt.ylabel('$x_n$', fontsize=11)
    plt.grid(True, alpha=0.3)

    # неподвижные точки
    if r > 0:
        sqrt_term = np.sqrt(r**2 + 4*r)
        x2 = (3*r - sqrt_term) / (2*r)
        x3 = (3*r + sqrt_term) / (2*r)

        if 0 <= x2 <= 2:
            plt.axhline(y=x2, color='r', linestyle='--', alpha=0.5, linewidth=0.8)
        if 0 <= x3 <= 2:
            plt.axhline(y=x3, color='r', linestyle='--', alpha=0.5, linewidth=0.8)

    plt.axhline(y=0, color='gray', linestyle=':', alpha=0.5, linewidth=0.5)
    plt.ylim(-0.1, 1.6)

# легенда
plt.subplot(2, 3, 6)
plt.axis('off')
plt.text(0.1, 0.7, 'Обозначения:', fontsize=11)
plt.text(0.1, 0.6, '• Сплошная линия: $x_n$', fontsize=10)
plt.text(0.1, 0.5, '• Красный пунктир: неподвижные точки', fontsize=10)
plt.text(0.1, 0.3, '$g(x) = rx(1-x)(2-x)$', fontsize=11)
plt.text(0.1, 0.2, f'$x_0 = {x0}$', fontsize=10)

plt.tight_layout()
plt.show()
```

# Hard level

## Задание 1: Циклы при $r \in (3; r_\infty)$

### 1.1 Изменение длины цикла при $r \in (3; r_\infty)$

**Анализ:**  
При $r > 3$ в логистическом отображении $x_{n+1} = rx_n(1-x_n)$ устойчивая неподвижная точка $x^* = 1 - 1/r$ теряет устойчивость. Вместо сходимости к одной точке появляются **циклы** — последовательности, которые повторяются через несколько шагов.

**Последовательность бифуркаций:**
- При $r \approx 3.0$: появляется **цикл периода 2** ($m = 2$)
- При $r \approx 3.44949$: цикл периода 2 удваивается до **цикла периода 4** ($m = 4$)
- При $r \approx 3.54409$: цикл периода 4 удваивается до **цикла периода 8** ($m = 8$)
- При $r \approx 3.56441$: цикл периода 8 удваивается до **цикла периода 16** ($m = 16$)
- При $r \to r_\infty \approx 3.56995$: длина цикла стремится к бесконечности

**Ответ:** При увеличении $r$ от 3 до $r_\infty$ длина цикла $m$ последовательно удваивается: $2 \to 4 \to 8 \to 16 \to 32 \to \dots$

### 1.2 Ограничения на длину цикла $m$

**Экспериментальный анализ показывает:**
При $r \in (3; r_\infty)$ длина цикла $m$ может принимать **только значения, являющиеся степенями двойки**:


$m = 2^k \quad \text{где } k = 1, 2, 3, 4, \dots$

То есть: $m \in \{2, 4, 8, 16, 32, 64, \dots\}$

**Объяснение:**  
Это явление называется **каскадом бифуркаций удвоения периода**. Каждая следующая бифуркация удваивает период существующего цикла. Такая закономерность характерна для широкого класса одномерных отображений и является универсальным свойством перехода к хаосу.

**Доказательство идеи:**  
Рассмотрим второе итерирование $f^{(2)}(x) = f(f(x))$. При $r > 3$ у $f^{(2)}(x)$ появляются две новые неподвижные точки, которые соответствуют циклу периода 2 исходного отображения. При дальнейшем увеличении $r$ у $f^{(4)}(x) = f^{(2)}(f^{(2)}(x))$ появляются новые неподвижные точки, что соответствует циклу периода 4, и т.д.

## Задание 2: Лестница Ламерея

### 2.1 Метод построения

Лестница Ламерея — графический метод визуализации сходимости к неподвижной точке:

1. Начинаем с точки $(x_0, 0)$
2. Двигаемся вертикально к графику отображения: $(x_0, x_1)$ где $x_1 = f(x_0)$
3. Двигаемся горизонтально к линии $y = x$: $(x_1, x_1)$
4. Повторяем: вертикально к $(x_1, x_2)$, горизонтально к $(x_2, x_2)$, и т.д.



```python
def plot_lamey_staircase(r, x0=0.3, iterations=20):
    """строит лестницу Ламерея для логистического отображения"""

    # генерируем последовательность
    x_vals = [x0]
    for n in range(iterations):
        x_next = r * x_vals[-1] * (1 - x_vals[-1])
        x_vals.append(x_next)

    # строим лестницу Ламерея
    plt.figure(figsize=(8, 8))

    # график функции
    x = np.linspace(0, 1, 400)
    y = r * x * (1 - x)
    plt.plot(x, y, 'b-', linewidth=2, label=f'$f(x) = {r}x(1-x)$')

    # линия y = x
    plt.plot(x, x, 'k--', alpha=0.5, label='$y = x$')

    # строим лестницу
    for i in range(iterations):
        # вертикальный отрезок: (x_i, x_i) → (x_i, x_{i+1})
        plt.plot([x_vals[i], x_vals[i]],
                 [x_vals[i], x_vals[i+1]],
                 'r-', alpha=0.7, linewidth=1)

        # горизонтальный отрезок: (x_i, x_{i+1}) → (x_{i+1}, x_{i+1})
        plt.plot([x_vals[i], x_vals[i+1]],
                 [x_vals[i+1], x_vals[i+1]],
                 'r-', alpha=0.7, linewidth=1)

    plt.plot(x0, 0, 'go', markersize=8, label=f'Начальная точка: $x_0={x0}$')

    x_star = 1 - 1/r if r > 1 else 0
    plt.plot(x_star, x_star, 'ro', markersize=8, label=f'Неподвижная точка: {x_star:.3f}')

    plt.title(f'Лестница Ламерея для r = {r}', fontsize=14)
    plt.xlabel('$x_n$', fontsize=12)
    plt.ylabel('$x_{n+1}$', fontsize=12)
    plt.grid(True, alpha=0.3)
    plt.legend(fontsize=10)
    plt.xlim(0, 1)
    plt.ylim(0, 1)
    plt.axis('equal')

    plt.tight_layout()
    plt.show()

print("Примеры лестницы Ламерея для разных r:")

print("1. r = 2.5 (сходимость к неподвижной точке)")
plot_lamey_staircase(r=2.5, x0=0.2, iterations=10)

print("\n2. r = 3.2 (цикл периода 2)")
plot_lamey_staircase(r=3.2, x0=0.2, iterations=20)

print("\n3. r = 3.5 (более сложное поведение)")
plot_lamey_staircase(r=3.5, x0=0.2, iterations=30)
```

### 2.2 Выводы о внешнем виде циклов на графике

1. **Неподвижная точка ($m = 1$)**:
   - Лестница "закручивается" по спирали к точке пересечения $f(x)$ и $y = x$
   - С каждым шагом отрезки становятся короче

2. **Цикл периода 2 ($m = 2$)**:
   - Лестница образует прямоугольник
   - Два вертикальных и два горизонтальных отрезка повторяются
   - Точки цикла находятся на пересечении $f^{(2)}(x)$ с $y = x$

3. **Цикл периода 4 ($m = 4$)**:
   - Образуется фигура из 8 отрезков (4 вертикальных + 4 горизонтальных)
   - Лестница проходит через 4 точки цикла
   - Видна более сложная структура

4. **Циклы большего периода**:
   - Образуют сложные многоугольники
   - С увеличением периода фигура становится более запутанной

5. **Хаотический режим ($r > r_\infty$)**:
   - Лестница заполняет область без явного порядка
   - Отрезки имеют разную длину и не образуют регулярных фигур

## Задание 3: Исследование отображения $g(x) = rx(1-x)(2-x)$

### 3.1 Исследование изменения длины цикла


```python
import numpy as np
import matplotlib.pyplot as plt

def g(x, r):
    return r * x * (1 - x) * (2 - x)

# График 1: Сравнение функций
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
x = np.linspace(0, 1, 400)
for r in [0.5, 1.0, 1.5, 2.0, 2.5]:
    y = g(x, r)
    plt.plot(x, y, label=f'r = {r}')

plt.plot(x, x, 'k--', alpha=0.5)
plt.title('отображение g(x) = rx(1-x)(2-x)', fontsize=12)
plt.xlabel('x', fontsize=11)
plt.ylabel('g(x)', fontsize=11)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=9)
plt.ylim(-0.1, 1.6)

# График 2: примеры последовательностей
plt.subplot(1, 2, 2)
r_values = [0.3, 1.0, 2.0, 2.5]
x0 = 0.3

for r in r_values:
    x_vals = [x0]
    for _ in range(30):
        x_vals.append(g(x_vals[-1], r))

    plt.plot(x_vals, 'o-', markersize=3, label=f'r = {r}')

plt.title('Последовательности для разных r', fontsize=12)
plt.xlabel('Номер итерации n', fontsize=11)
plt.ylabel('x_n', fontsize=11)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=9)

plt.tight_layout()
plt.show()

# График 3: Области притяжения
plt.figure(figsize=(8, 6))

r_range = np.linspace(0.5, 2.5, 300)
x = 0.3
collect = 50

for r in r_range:
    for _ in range(200):
        x = g(x, r)

    points = []
    for _ in range(collect):
        x = g(x, r)
        points.append((r, x))

    for r_point, x_point in points:
        plt.plot(r_point, x_point, ',k', alpha=0.3)

plt.title('Бифуркационная диаграмма для g(x) = rx(1-x)(2-x)', fontsize=12)
plt.xlabel('Параметр r', fontsize=11)
plt.ylabel('Установившиеся значения x_n', fontsize=11)
plt.grid(True, alpha=0.2)
plt.axvline(x=3*np.sqrt(3)/2, color='r', linestyle='--',
           label=f'r_max = {3*np.sqrt(3)/2:.3f}')
plt.legend()

plt.tight_layout()
plt.show()
```


**Аналитический подход:**

1. **Неподвижные точки:** Решаем $x = g(x) = rx(1-x)(2-x)$
   - $x_1^* = 0$ (всегда)
   - $x_{2,3}^* = \frac{3r \pm \sqrt{r^2 + 4r}}{2r}$ (при $r > 0$)

2. **Устойчивость точки $x=0$:**  
   $g'(x) = r(2 - 6x + 3x^2)$, $g'(0) = 2r$  
   Точка $x=0$ устойчива при $|2r| < 1$, т.е. $r < 0.5$

3. **Появление циклов:**  
   - При $r \approx 0.5$: точка $x=0$ теряет устойчивость
   - При $r \approx 1.2-1.5$: появляются первые циклы
   - При $r > 2.0$: сложная динамика, возможны длинные циклы

**Качественный анализ:**  
Точечное отображение, будучи кубическим, имеет более богатую динамику, чем квадратичное логистическое отображение. Оно может иметь до трех неподвижных точек и более сложные бифуркационные сценарии.

### 3.2 Сравнение с логистическим отображением

**Сходства:**
1. Оба отображения демонстрируют бифуркации при увеличении $r$
2. В обоих случаях возможен переход к хаотическому поведению
3. Существуют области параметров с циклами конечной длины

**Различия:**
1. **Диапазон параметра:** Логистическое: $r \in [0, 4]$, Ваше: $r \in [0, \frac{3\sqrt{3}}{2} \approx 2.598]$
2. **Структура бифуркаций:**  
   - Логистическое: четкий каскад удвоения периода $2^k$
   - Точечное: возможны циклы различных длин, не только степени двойки
3. **Сложность динамики:** Кубическое отображение может давать более разнообразное поведение при меньших значениях $r$
4. **Количество неподвижных точек:** Логистическое — до 2, точечное — до 3

**Вывод:** Хотя точечное отображение структурно сложнее логистического, оно демонстрирует аналогичные качественные явления: бифуркации, циклы, переход к хаосу. Однако конкретные значения параметров, при которых происходят эти переходы, и структура возникающих циклов различны.
